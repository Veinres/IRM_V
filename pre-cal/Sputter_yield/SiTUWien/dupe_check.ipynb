```python
import pandas as pd

df = pd.read_csv("vienna_SiSi.csv", header=None)

# how many duplicate energies in column 0 (the eV axis)
dupes = len(df[0]) - df[0].nunique()
print("duplicate energies:", dupes)

# any NaN in the energy column
print("NaNs in energy col:", df[0].isna().sum())

# show the actual duplicate rows, if any
print(df[df[0].duplicated(keep=False)])
```
